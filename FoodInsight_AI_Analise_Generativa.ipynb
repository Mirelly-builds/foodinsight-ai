# FoodInsight AI
### Análise de avaliações e recomendações inteligentes para restaurantes

**Projeto-piloto | Cliente fictício: Momoshi**

**Objetivo:** desenvolver um protótipo que utiliza inteligência artificial generativa para analisar avaliações de clientes, identificar sentimentos e categorias de problemas e produzir recomendações de melhoria.

**Tecnologias utilizadas:**
- Python e Google Colab
- Modelo de IA generativa via OpenRouter
- Pandas para organização e processamento dos dados
- Excel para validação dos resultados
- Looker Studio para visualização dos indicadores

**Escopo:** análise de 50 avaliações fictícias, com classificação, justificativas e recomendações geradas por IA, submetidas à revisão humana.



## 1. Problema identificado

Restaurantes recebem avaliações constantemente em diferentes canais digitais. Entretanto, transformar esse grande volume de comentários em informações organizadas para apoiar decisões pode ser um desafio.

Comentários sobre qualidade dos alimentos, entrega, atendimento e pedidos incorretos podem revelar problemas recorrentes, mas nem sempre são analisados de forma estruturada.

O FoodInsight AI foi proposto para explorar como a inteligência artificial generativa pode ajudar nesse processo, transformando opiniões dos clientes em classificações, justificativas e recomendações de melhoria.



## 2. Metodologia e fluxo de funcionamento

O desenvolvimento do FoodInsight AI foi dividido em etapas:

1. **Preparação dos dados:** organização das 50 avaliações fictícias do restaurante Momoshi.
2. **Processamento:** utilização de Python e Pandas para estruturar e preparar os registros.
3. **Análise generativa:** envio dos comentários a um modelo de IA generativa para interpretação.
4. **Classificação:** identificação de sentimentos e categorias relacionadas às experiências dos clientes.
5. **Geração de recomendações:** produção de justificativas e sugestões de melhoria com base nas avaliações.
6. **Validação humana:** revisão das respostas geradas pela IA e ajustes necessários.
7. **Visualização:** organização dos resultados em um dashboard interativo no Looker Studio.

O fluxo foi desenvolvido como um protótipo experimental, com participação humana na validação dos resultados.



## 3. Tecnologias utilizadas

**Python e Google Colab:** utilizados para executar o processamento dos dados e organizar o fluxo de análise.

**Pandas:** utilizado na manipulação, organização e validação dos registros.

**OpenRouter:** utilizado como intermediário para acessar um modelo de inteligência artificial generativa.

**IA generativa:** responsável por interpretar os comentários, identificar sentimentos e categorias, apresentar justificativas e sugerir recomendações.

**Excel:** utilizado para organizar e conferir os resultados produzidos.

**Looker Studio:** utilizado para construir o painel interativo, permitindo visualizar indicadores, sentimentos, categorias e recomendações.

### Arquitetura simplificada

Avaliações → Python e Pandas → Modelo de IA generativa → Revisão humana → Base estruturada → Dashboard interativo.



## 4. Resultados obtidos

O processamento das 50 avaliações fictícias do Momoshi gerou os seguintes resultados:

- **50 avaliações** analisadas.
- **50 análises generativas** realizadas.
- **50 recomendações finais** produzidas.
- **8 recomendações** passaram por ajustes após revisão humana.

### Distribuição dos sentimentos

- Negativas: 19 (38%).
- Mistas: 17 (34%).
- Positivas: 14 (28%).

### Categorias identificadas

- Qualidade dos alimentos: 23 ocorrências.
- Entrega: 9 ocorrências.
- Atendimento: 9 ocorrências.
- Pedido correto: 8 ocorrências.
- Embalagem: 6 ocorrências.
- Preço: 4 ocorrências.
- Experiência: 2 ocorrências.
- Variedade: 1 ocorrência.

As categorias podem aparecer em conjunto em uma mesma avaliação. Por isso, a soma das ocorrências por categoria é superior ao número de avaliações analisadas.

Os resultados demonstram como a IA generativa pode apoiar a organização de opiniões dos clientes e a identificação de oportunidades de melhoria.



## 5. Validação humana e limitações

### Validação humana

As respostas produzidas pela IA generativa foram revisadas para verificar a coerência das classificações, justificativas e recomendações.

Durante essa etapa, 8 das 50 recomendações precisaram de ajustes humanos.

Essa revisão foi importante para melhorar a qualidade das sugestões e evitar que respostas inadequadas fossem consideradas conclusões definitivas.

### Limitações do protótipo

- Utilização de avaliações fictícias, sem dados reais de clientes.
- Base experimental limitada a 50 avaliações.
- Possibilidade de interpretações incorretas ou inconsistentes pelo modelo de IA.
- Necessidade de revisão humana das recomendações.
- Ausência de integração contínua com plataformas reais de restaurantes.
- Resultados ainda não validados em ambiente operacional.

O projeto demonstra uma aplicação experimental de IA generativa, não uma solução pronta para implantação comercial.



## 6. Conclusão e próximos passos

O FoodInsight AI permitiu explorar o uso da inteligência artificial generativa na análise de avaliações de clientes, transformando comentários em informações estruturadas e recomendações de melhoria.

O desenvolvimento integrou processamento de dados, IA generativa, validação humana e visualização de informações em um dashboard interativo.

Além do resultado técnico, o projeto proporcionou aprendizado prático sobre a aplicação de IA generativa em um problema de negócio.

### Possibilidades de evolução

- Testar o protótipo com avaliações reais, mediante autorização e cuidados com dados pessoais.
- Ampliar a base de dados para identificar padrões mais consistentes.
- Aprimorar os critérios de classificação e validação das respostas.
- Avaliar a integração com plataformas de pedidos e avaliações.
- Desenvolver mecanismos de atualização automatizada dos indicadores.

**Proposta de valor:** transformar a voz do cliente em informações que apoiem decisões mais inteligentes na gestão de restaurantes.



````python

import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

print("FoodInsight AI iniciado!")
print("Bibliotecas carregadas com sucesso.")
````

````python

# Enviar nossa planilha
uploaded = files.upload()

# Identificar o arquivo enviado
arquivo = list(uploaded.keys())[0]

# Ler a aba de avaliações
df = pd.read_excel(arquivo, sheet_name="Avaliacoes")

print("Planilha importada com sucesso!")
````

````python

# Mostrar as cinco primeiras avaliações
display(df.head())

# Verificar quantidade de registros e colunas
print("Total de avaliações:", len(df))
print("Total de colunas:", len(df.columns))

# Calcular a nota média
print("Nota média:", df["Nota (1 a 5)"].mean())

# Verificar dados vazios
print("\nDados vazios:")
print(df.isnull().sum())
````

````python

# Contagem dos sentimentos
sentimentos = df["Sentimento (IA)"].value_counts()

print("DISTRIBUIÇÃO DOS SENTIMENTOS")
print(sentimentos)

# Calcular os percentuais
percentuais = df["Sentimento (IA)"].value_counts(normalize=True) * 100

print("\nPERCENTUAL POR SENTIMENTO")
print(percentuais.round(1))
````

````python

# Separar categorias múltiplas
categorias = (
    df["Categoria (IA)"]
    .dropna()
    .str.split(";")
    .explode()
    .str.strip()
)

# Contar ocorrências
contagem_categorias = categorias.value_counts()

print("PROBLEMAS MAIS MENCIONADOS")
print(contagem_categorias)
````

````python

plt.figure(figsize=(9, 5))

contagem_categorias.plot(
    kind="bar",
    title="Principais categorias das avaliações - Momoshi",
    xlabel="Categoria",
    ylabel="Quantidade de ocorrências"
)

plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()
````

````python

# Contar quantas vezes cada nota apareceu
notas = df["Nota (1 a 5)"].value_counts().sort_index()

print("DISTRIBUIÇÃO DAS NOTAS")
print(notas)

# Calcular a média
media = df["Nota (1 a 5)"].mean()

print("\nNota média do Momoshi:", round(media, 2))
````

````python

# Cruzar notas e sentimentos
diagnostico = pd.crosstab(
    df["Nota (1 a 5)"],
    df["Sentimento (IA)"]
)

print("RELAÇÃO ENTRE NOTAS E SENTIMENTOS")
display(diagnostico)
````

````python

plt.figure(figsize=(8, 5))

notas.plot(
    kind="bar",
    title="Distribuição das notas dos clientes - Momoshi",
    xlabel="Nota",
    ylabel="Quantidade de avaliações"
)

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
````

````python

# Separar as categorias das avaliações
dados_categorias = df.copy()

dados_categorias["Categoria"] = (
    dados_categorias["Categoria (IA)"]
    .fillna("")
    .str.split(";")
)

dados_categorias = dados_categorias.explode("Categoria")

dados_categorias["Categoria"] = (
    dados_categorias["Categoria"].str.strip()
)

# Cruzar categorias e sentimentos
analise_problemas = pd.crosstab(
    dados_categorias["Categoria"],
    dados_categorias["Sentimento (IA)"]
)

# Ordenar pelas ocorrências negativas
analise_problemas = analise_problemas.sort_values(
    by="Negativo",
    ascending=False
)

print("DIAGNÓSTICO POR CATEGORIA")
display(analise_problemas)
````

````python

# Somar avaliações negativas e mistas
analise_problemas["Possivel_atencao"] = (
    analise_problemas["Negativo"] +
    analise_problemas["Misto"]
)

# Ordenar por quantidade de menções que merecem atenção
ranking = analise_problemas.sort_values(
    by="Possivel_atencao",
    ascending=False
)

print("RANKING DE CATEGORIAS QUE MERECEM ATENÇÃO")
display(ranking)
````

````python

print("FOODINSIGHT AI")
print("DIAGNÓSTICO DO RESTAURANTE MOMOSHI")
print("-" * 45)

total = len(df)
negativas = (df["Sentimento (IA)"] == "Negativo").sum()
mistas = (df["Sentimento (IA)"] == "Misto").sum()

media = df["Nota (1 a 5)"].mean()

print(f"Total de avaliações: {total}")
print(f"Nota média: {media:.2f}")
print(f"Avaliações negativas: {negativas} ({negativas/total*100:.1f}%)")
print(f"Avaliações mistas: {mistas} ({mistas/total*100:.1f}%)")

print("\nPRINCIPAIS PONTOS DE ATENÇÃO:")

for categoria, linha in ranking.head(3).iterrows():
    print(
        f"- {categoria}: "
        f"{int(linha['Possivel_atencao'])} menções "
        f"negativas ou mistas."
    )
````

````python

# FOODINSIGHT AI - MOTOR DE RECOMENDACOES

recomendacoes = {
    "Qualidade dos alimentos":
        "Revisar os padrões de preparo, conservação e apresentação dos alimentos.",

    "Entrega":
        "Analisar os atrasos e melhorar o planejamento e acompanhamento das entregas.",

    "Pedido incorreto":
        "Implementar uma conferência dos itens antes de liberar cada pedido.",

    "Atendimento":
        "Avaliar o atendimento e padronizar os procedimentos de comunicação com clientes.",

    "Embalagem":
        "Revisar a vedação e a resistência das embalagens durante o transporte.",

    "Preço/custo-benefício":
        "Avaliar a percepção de valor dos clientes em relação aos preços praticados."
}

print("=" * 55)
print("FOODINSIGHT AI | PLANO DE MELHORIA DO MOMOSHI")
print("=" * 55)

for categoria, linha in ranking.head(5).iterrows():

    quantidade = int(linha["Possivel_atencao"])

    if categoria in recomendacoes:
        print(f"\nCATEGORIA: {categoria}")
        print(f"OCORRENCIAS: {quantidade}")
        print(f"RECOMENDACAO: {recomendacoes[categoria]}")
        print("-" * 55)
````

````python

# FOODINSIGHT AI - RELATORIO DE MELHORIA

relatorio = []

for categoria, linha in ranking.iterrows():
    relatorio.append({
        "Categoria": categoria,
        "Negativas": int(linha["Negativo"]),
        "Mistas": int(linha["Misto"]),
        "Positivas": int(linha["Positivo"]),
        "Total de atencao": int(linha["Possivel_atencao"]),
        "Recomendacao": recomendacoes.get(
            categoria,
            "Investigar a categoria."
        )
    })

df_relatorio = pd.DataFrame(relatorio)

# Exportar resultados
with pd.ExcelWriter("Relatorio_FoodInsight_Momoshi.xlsx") as writer:
    df.to_excel(writer, sheet_name="Avaliacoes", index=False)
    df_relatorio.to_excel(writer, sheet_name="Plano de Melhoria", index=False)
    diagnostico.to_excel(writer, sheet_name="Notas e Sentimentos")

print("Relatorio gerado com sucesso!")

files.download("Relatorio_FoodInsight_Momoshi.xlsx")
````

````python
def analisar_comentario(comentario):

    payload = {
        "model": "openrouter/free",
        "messages": [
            {
                "role": "system",
                "content": """
                Você é um analista de experiência do cliente de restaurantes.
                Analise o comentário e responda em português com:
                1. Sentimento: Positivo, Negativo ou Misto.
                2. Categoria principal.
                3. Justificativa.
                4. Recomendação prática.
                """
            },
            {
                "role": "user",
                "content": f"Analise este comentário: {comentario}"
            }
        ]
    }

    resposta = requests.post(
        url,
        headers=headers,
        json=payload,
        timeout=60
    )

    if resposta.status_code == 200:
        return resposta.json()["choices"][0]["message"]["content"]

    return f"Erro {resposta.status_code}: {resposta.text[:500]}"
````

````python
import json

amostra = df[["ID", "Comentário do cliente"]].head(5)

comentarios = amostra.to_dict(orient="records")

payload = {
    "model": "openrouter/free",
    "messages": [
        {
            "role": "system",
            "content": """
            Analise cada avaliação de restaurante individualmente.
            Responda somente em JSON válido, como uma lista de objetos.
            Cada objeto deve conter:
            ID, sentimento, categoria, justificativa e recomendacao.
            O sentimento deve ser Positivo, Negativo ou Misto.
            Não invente informações.
            """
        },
        {
            "role": "user",
            "content": json.dumps(comentarios, ensure_ascii=False)
        }
    ]
}

response = requests.post(url, headers=headers, json=payload)

if response.status_code == 200:
    print(response.json()["choices"][0]["message"]["content"])
else:
    print("Erro:", response.status_code)
    print(response.text[:1000])
````

````python
import re

conteudo = response.json()["choices"][0]["message"]["content"].strip()

conteudo = re.sub(r"^```(?:json)?\s*|\s*```$", "", conteudo)

if not conteudo.startswith("["):
    conteudo = "[" + conteudo

if not conteudo.endswith("]"):
    conteudo += "]"

try:
    resultados_ia = json.loads(conteudo)
    df_ia = pd.DataFrame(resultados_ia)

    print("Análises organizadas com sucesso!")
    display(df_ia)

except json.JSONDecodeError as erro:
    print("A resposta precisa de ajuste:", erro)
    print(conteudo)
````

````python
# FREQUÊNCIA DAS CATEGORIAS IDENTIFICADAS PELA IA

frequencia = df_ia["categoria"].value_counts().reset_index()

frequencia.columns = ["Categoria identificada", "Quantidade"]

display(frequencia)
````

````python
# FOODINSIGHT AI - PADRONIZAÇÃO DE CATEGORIAS

regras = {
    "Qualidade dos alimentos": [
        "qualidade", "alimento", "comida", "sabor",
        "porção", "apresentação", "temperatura",
        "preparação", "preparo", "composição",
        "segurança alimentar",
    ],
    "Entrega": [
        "entrega", "prazo", "tempo de entrega"
    ],
    "Pedido correto": [
        "pedido incompleto", "pedido incorreto",
        "acurácia do pedido", "completude do pedido",
        "conformidade do pedido", "pedido e entrega",
        "conformidade", "restrição alimentar"
    ],
    "Embalagem": [
        "embalagem", "conservação", "integridade"
    ],
    "Atendimento": [
        "atendimento", "suporte", "comunicação"
    ],
    "Preço/custo-benefício": [
        "preço", "custo-benefício", "cobrança"
    ],
    "Variedade do cardápio": [
        "variedade", "opções", "cardápio"
    ],
    "Experiência geral": [
        "experiência", "satisfação geral"
    ]
}
def padronizar_categorias(valor):
    texto = str(valor).casefold()

    categorias = []

    for categoria, palavras in regras.items():
        if any(palavra in texto for palavra in palavras):
            categorias.append(categoria)

    if not categorias:
        categorias.append("Revisão manual")

    return categorias

df_ia["Categorias padronizadas"] = (
    df_ia["categoria"].apply(padronizar_categorias)
)

display(
    df_ia[["ID", "categoria", "Categorias padronizadas"]].head(10)
)
````

````python
# CONTAGEM DAS CATEGORIAS PADRONIZADAS

contagem_categorias = (
    df_ia["Categorias padronizadas"]
    .explode()
    .value_counts()
    .reset_index()
)

contagem_categorias.columns = ["Categoria", "Quantidade"]

display(contagem_categorias)

print("Total de avaliações:", len(df_ia))
print("Total de classificações:", contagem_categorias["Quantidade"].sum())
````

````python
def gerar_recomendacao_revisada(comentario):
    payload = {
        "model": "openrouter/free",
        "messages": [
            {
                "role": "system",
                "content": """
Você é um consultor de experiência do cliente especializado em restaurantes.

Sua tarefa é recomendar ações práticas para o GESTOR do restaurante,
com base exclusivamente no comentário do cliente.

Regras:
- Escreva uma recomendação em português.
- Use linguagem profissional, clara e objetiva.
- Sugira uma ação que a equipe ou gestão possa executar.
- Não dê conselhos ao consumidor.
- Não mande o cliente voltar, escolher outro prato ou evitar produtos.
- Não culpe funcionários ou entregadores sem evidências.
- Não invente causas ou fatos que não aparecem no comentário.
- Em reclamações sobre porções, recomende verificar e padronizar
  as quantidades antes de sugerir aumentos.
- Em reclamações sobre alimentos, considere controles de qualidade,
  armazenamento, preparo e conservação, conforme o caso.
- Em avaliações positivas, recomende manter os padrões responsáveis
  pela boa experiência.
- Limite a resposta a uma ou duas frases.

Retorne somente a recomendação, sem título ou explicação.
"""
            },
            {
                "role": "user",
                "content": f"Comentário do cliente: {comentario}"
            }
        ]
    }

    resposta = requests.post(
        "https://openrouter.ai/api/v1/chat/completions",
        headers={
            "Authorization": f"Bearer {userdata.get('OPENROUTER_API_KEY')}",
            "Content-Type": "application/json"
        },
        json=payload
    )

    resposta.raise_for_status()

    return resposta.json()["choices"][0]["message"]["content"].strip()
````
